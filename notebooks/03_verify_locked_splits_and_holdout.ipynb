# 03 — Verify Locked Splits and Holdout
Verifies train/validation/test counts and 50 real + 50 fake holdout. Holdout labels come from folder names.

```python
from pathlib import Path
import sys
ROOT=Path.cwd()
while ROOT!=ROOT.parent and not (ROOT/"data").exists(): ROOT=ROOT.parent
if str(ROOT) not in sys.path: sys.path.insert(0,str(ROOT))
print("PROJECT_ROOT:",ROOT)
DATASET=ROOT/"data"/"raw"/"deepfake_merged_dataset"
print("DATASET:",DATASET)
assert DATASET.exists(), f"Missing dataset: {DATASET}"
from src.data_utils import iter_images
expected={"train/real":102358,"train/fake":128003,"val/real":19853,"val/fake":28436,"test/real":11199,"test/fake":14305}
for rel,n in expected.items(): assert sum(1 for _ in iter_images(DATASET/rel))==n
H=ROOT/"data/holdout_50"
assert sum(1 for _ in iter_images(H/"real"))==50
assert sum(1 for _ in iter_images(H/"fake"))==50
locked={str(p.resolve()) for s in ["train","val","test"] for p in iter_images(DATASET/s)}
hold={str(p.resolve()) for c in ["real","fake"] for p in iter_images(H/c)}
assert locked.isdisjoint(hold)
print("LOCKED SPLITS + 100-IMAGE HOLDOUT GATE PASSED")
```
