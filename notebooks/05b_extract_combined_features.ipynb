# 05b — Build Combined Features
**Does not process images.** Combines the existing normal CSV with the all-forensic CSV by filepath and label.

```python
from pathlib import Path
import sys
ROOT=Path.cwd()
while ROOT!=ROOT.parent and not (ROOT/"data").exists(): ROOT=ROOT.parent
if str(ROOT) not in sys.path: sys.path.insert(0,str(ROOT))
print("PROJECT_ROOT:",ROOT)
DATASET=ROOT/"data"/"raw"/"deepfake_merged_dataset"
print("DATASET:",DATASET)
assert DATASET.exists(), f"Missing dataset: {DATASET}"
import pandas as pd
from src.features import NORMAL_COLS,ALL_COLS
for split in ["train","val"]:
    n=pd.read_csv(ROOT/"data/processed"/f"{split}_normal_features.csv")
    a=pd.read_csv(ROOT/"data/processed"/f"{split}_all_forensic_features.csv")
    # The all-forensic file already contains the original normal features.
    assert n[["filepath","label"]].equals(a[["filepath","label"]])
    assert all(c in a.columns for c in ALL_COLS)
    a.to_csv(ROOT/"data/processed"/f"{split}_combined_features.csv",index=False)
    print(split,a.shape)
```
